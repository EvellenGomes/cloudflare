```python
import time
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python

url = 'https://www.scrapingcourse.com/login/cf-antibot' 
email = 'admin@example.com'
password = 'password'


options = uc.ChromeOptions()
driver = uc.Chrome(options=options)

print("Acessando a página...")
try:
    driver.get(url)
    print("Quebrando Cloudflare")
    wait = WebDriverWait(driver, 30) 
    
    email_input = wait.until(
        EC.presence_of_element_located((By.ID, 'email'))
    )
    print("Página de login carregada.")


    password_input = driver.find_element(By.ID, 'password')
    login_button = driver.find_element(By.CSS_SELECTOR, 'button[type="submit"]')


    print(f"Preenchendo o e-mail: {email}")
    email_input.send_keys(email)
    print("Preenchendo a senha...")
    password_input.send_keys(password)
    print("Clicando no botão de login..")
    login_button.click()

   
    print("\nLogin realizado com sucesso!")
    
    products_header = wait.until(
        EC.presence_of_element_located((By.XPATH, "//h2[contains(text(), 'Products')]"))
    )
    print(f"Página de produtos carregada...'")
    

    time.sleep(10)

except Exception as e:
    print(f"\nOcorreu um erro durante o processo: {e}")

finally:
    print("\nFechando o navegador.")
    driver.quit()
```
